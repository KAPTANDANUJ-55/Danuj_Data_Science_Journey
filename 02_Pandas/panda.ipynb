## What is Pandas?

**Pandas** is an open-source Python library built for handling, cleaning, and analyzing tabular data. Think of it as **Excel inside Python**, except it can process millions of rows smoothly and automate repetitive tasks with a few lines of code.

### Core Building Blocks
* **Series:** A one-dimensional labeled array (like a single column in a spreadsheet).
* **DataFrame:** A two-dimensional table with labeled rows and columns (like a full Excel sheet or SQL table).

### What You Can Do With It
* **Import & Export:** Read and write data across CSV, Excel, JSON, Parquet, and SQL databases.
* **Clean Data:** Quickly find, fill, or drop missing values (`NaN`) and fix messy data types.
* **Filter & Sort:** Query millions of rows using simple conditions (e.g., `df[df["sales"] > 1000]`).
* **Group & Aggregate:** Summarize data with pivot tables, `.groupby()`, averages, and counts.
* **Merge & Join:** Combine multiple datasets together just like SQL joins or `VLOOKUP`.


```python
import pandas as pd

# Series
s = pd.Series([10, 20, 30], index=["a", "b", "c"])
print(s["b"])   # 20

# DataFrame
df = pd.DataFrame({
    "Name": ["Ravi", "Amit", "Neha", "Priya"],
    "Age": [21, 22, 20, 23],
    "Marks": [78, 85, 92, 66]
})
print(df)
```
