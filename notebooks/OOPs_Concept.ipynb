# OOPs Concepts: Classes

```python
class Phone:
    brand = "Xiaomi"        # Class Variable
    number_of_phone_users = 0;


    def __init__(self,model,price):
        self.model = model         # instance variable
        self.price = price
        Phone.number_of_phone_users +=1


    def info(self):
        return f"{self.brand} {self.model} - Rs.{self.price}"
            
p1 = Phone("17T","69,999")
p2 = Phone("Redmi Note 17 Pro Max 5G","49,999")
p3 = Phone("14","61,900")
p4 = Phone("Poco F8","55,778")
p5 = Phone("Redmi Turbo 5","41,999")


print(p1.info())
print(p2.info())

print(p1.brand, p2.brand)      # Samsung Samsung
Phone.brand = "Apple"          # Changed With Class
print(p1.brand, p2.brand)      # Apple Apple (both have effect)

p1.brand = "Nokia"             # only p1 brand is changed
print(p1.brand, p2.brand)      # Nokia Apple

print(p1.__dict__)   # because we explicit change the p1 so brand comes as new key value pair
print(p2.__dict__)   # because we did not explicitly changed the brand for p2

# if we supossed to do

p2.brand = "Motorola"

print(p2.__dict__) 

# if i delete p1 brand

del p1.brand
print(p1.__dict__) 
print(p1.brand)


print(f"There are {Phone.number_of_phone_users} users of Xiaomi")
```
