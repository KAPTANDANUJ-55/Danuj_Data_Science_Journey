```python
import numpy as np
from numpy.lib.stride_tricks import as_strided

arr = np.array([[1, 2, 3], [4, 5, 6]])
# ========== Inbuilt Add-ons
print("Array:\n", arr)
print("Shape:", arr.shape)                    # (2, 3)
print("Number of dimensions:", arr.ndim)      # 2
print("Size (number of elements):", arr.size) # 6
print("Data type:", arr.dtype)                # int64 (Depends On Platform)
print("Item size (in bytes):", arr.itemsize)  # 8

print("Total bytes:", arr.nbytes)             # 48 (size × itemsize)
print("Strides:", arr.strides)                # (24, 8)
print("Transpose:\n", arr.T)                  # rows ↔ columns
print("Reshaped (3,2):\n", arr.reshape(3, 2))
print("Flattened:", arr.flatten())            # [1 2 3 4 5 6]
print("As float:", arr.astype(np.float32))    # dtype changed
print("As Strided Forcefully Stride: " , as_strided(arr,shape=(3,2), strides=(16,8)))
```

```python
import numpy as np

### numpy vectorized operation
arr1 = np.array([22,33,44,55,66])
arr2 = np.array((98,89,76,43))
print(arr2)

### Element Wise Addition
def safe_add(a, b):
    n = max(len(a), len(b))
    a = np.pad(a, (0, n - len(a)))
    b = np.pad(b, (0, n - len(b)))
    return a + b

def safe_sub(a, b):
    n = max(len(a), len(b))
    a = np.pad(a, (0, n - len(a)))
    b = np.pad(b, (0, n - len(b)))
    return a - b

def safe_mul(a, b):
    n = max(len(a), len(b))
    a = np.pad(a, (0, n - len(a)))
    b = np.pad(b, (0, n - len(b)))
    return a * b
# print("ADDITION: ", arr1+arr2) ## this will give value error because dimension is not same


print("ADDITION: ", safe_add(arr1,arr2)) ## Perfect !!!!
print("SUBSTRACTION: ", safe_sub(arr1,arr2))
print("MULTIPLICATION: ", safe_mul(arr1,arr2))

## YOU CAN ALSO DO NORMALLY LIKE a+b, a-b , a*b but that need same elements lets see



arr3 = np.array([17,21,56,877,20])
arr4 = np.array((77,88,27,86,45))

print("ADDITION: ", (arr3+arr4)) ## Perfect !!!!
print("SUBSTRACTION: ", (arr3-arr4))
print("MULTIPLICATION: ", (arr3*arr4))
```

```python
## Universal Function

arr = np.array([[3,4,5],[6,7,8]])

## sqrt function
print(np.sqrt(arr))
print(np.round(np.sqrt(arr), 2))

## exponenitial

print(np.exp(arr))


## sine
print(np.sin(arr))


## natural log
print(np.log(arr))

# log with base a(where a=1,2,3,4,5,6......∞)

print(np.log2(arr))
print(np.log10(arr))

## numpy has only log2 log10 loge or only log is exist
# print(np.log3(arr))
# print(np.log4(arr))

print(np.log(arr) / np.log(3))   # log base 3
print(np.log(arr) / np.log(4))   # log base 4

def log_base(x,base):
    return np.log(x)/np.log(base)

print(log_base(arr,5))    
```

```python
## statistical operations
import numpy as np

arr = np.array([[3, 4, 5], [6, 7, 8]])

# 1) Central tendency
print(np.mean(arr))    # 5.5
print(np.median(arr))  # 5.5
# mode is not available in NumPy; import it from SciPy:
# from scipy import stats; stats.mode(arr, axis=None)

# 2) Spread
print(np.min(arr), np.max(arr))  # 3 8
print(np.ptp(arr))               # 5 (max - min, range)
print(np.var(arr))               # 2.9167
print(np.std(arr))               # 1.7078

# 3) Sample vs Population
print(np.var(arr, ddof=1))  # 3.5
print(np.std(arr, ddof=1))  # 1.8708

# 4) Percentile / quartile
print(np.percentile(arr, 25))  # 4.25
print(np.percentile(arr, 50))  # 5.5
print(np.percentile(arr, 75))  # 6.75
print(np.quantile(arr, 0.75))  # 6.75 (same, on a 0–1 scale)

# 5) Position
print(np.argmin(arr), np.argmax(arr))  # 0 5 (flattened index)

# 6) Axis-wise
print(arr.mean(axis=0))  # [4.5 5.5 6.5] column-wise
print(arr.mean(axis=1))  # [4. 7.] row-wise

# 7) Correlation & covariance
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])
print(np.corrcoef(x, y))  # [[1.     0.7746]
                          #  [0.7746 1.    ]]
print(np.cov(x, y))       # [[2.5 1.5]
                          #  [1.5 1.5]]
```
